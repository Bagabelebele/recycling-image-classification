```python
from pathlib import Path

p = Path("src/ga_search.py")
code = p.read_text()

if '--trials' not in code:
    code = code.replace(
        'ap.add_argument("--out", default="results/ga")',
        'ap.add_argument("--out", default="results/ga")\n'
        '    ap.add_argument("--trials", type=int, default=POP * GENS)'
    )
    code = code.replace("range(POP * GENS)", "range(args.trials)")
    p.write_text(code)

print("Random-search budget is ready.")
```
